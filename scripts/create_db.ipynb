```python
# Скрипт для генерации датасета размером 30к строк

import pandas as pd
import numpy as np
import uuid
from datetime import datetime, timedelta

# Генерация данных (30 000 строк)
np.random.seed(42)
n = 30000

data = {
    "transaction_id": [str(uuid.uuid4()) for _ in range(n)],
    "user_id": np.random.randint(1000, 9999, n),
    "amount": np.round(np.random.uniform(10, 1000, n), 2),
    "currency": np.random.choice(["USD", "EUR", "RUB"], n),
    "transaction_date": [datetime.now() - timedelta(days=np.random.randint(1, 365)) for _ in range(n)],
    "status": np.random.choice(["completed", "pending", "failed"], n),
    "merchant": [f"merchant_{i}" for i in np.random.randint(1, 100, n)]
}

df = pd.DataFrame(data)

# Сохраняем в CSV
df.to_csv("transactions_v2.csv", index=False)
print("Файл transactions_v2.csv создан!")
```
