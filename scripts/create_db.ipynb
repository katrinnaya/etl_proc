```python
# Скрипт для генерации датасета размером 3,5к строк

import pandas as pd
import uuid
import random
from datetime import datetime, timedelta

def generate_transactions(num_rows=3500):
    currencies = ['RUB', 'USD', 'EUR']
    statuses = ['completed', 'pending', 'failed']
    merchants = [f'merchant_{i}' for i in range(1, 101)]
    start_date = datetime(2023, 1, 1)

    data = []
    for _ in range(num_rows):
        transaction_date = start_date + timedelta(
            days=random.randint(0, 365),
            hours=random.randint(0, 23),
            minutes=random.randint(0, 59),
            seconds=random.randint(0, 59)
        )

        data.append({
            'transaction_id': str(uuid.uuid4()),
            'user_id': random.randint(1000, 9999),
            'amount': round(random.uniform(10, 10000), 2),
            'currency': random.choice(currencies),
            'transaction_date': transaction_date.strftime('%Y-%m-%d %H:%M:%S.%f'),
            'status': random.choice(statuses),
            'merchant': random.choice(merchants)
        })

    return pd.DataFrame(data)

# Генерация и сохранение
df = generate_transactions() # Количество строк задаётся параметром num_rows по умолчанию
df.to_csv('transactions_v2.csv', index=False)
print("Файл transactions_v2.csv с 3500 строк успешно создан")
```
